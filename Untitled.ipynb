```python
!pip install pandas scikit-learn matplotlib
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("heart.csv")
```

```python
print(df.shape)        # rows, columns
```

```python
print(df.head())       # first 5 rows
```

```python
print(df.dtypes)       # column types
```

```python
print(df.isnull().sum()) # check missing values
```

```python
print(df["target"].value_counts()) # class balance
```

```python
# Age distribution by target
df.groupby("target")["age"].hist(alpha=0.6)
plt.legend(["No Disease", "Disease"])
plt.xlabel("Age")
plt.title("Age vs Heart Disease")
plt.show()
```

```python
import seaborn as sns
sns.heatmap(df.corr(), annot=True, fmt=".1f", cmap="coolwarm")
plt.show()
```

```python
X = df.drop("target", axis=1)  # all columns except target
y = df["target"]                # the column to predict

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,     # 20% for testing
    random_state=42    # fixed seed = reproducible
)

print("Train size:", X_train.shape)
print("Test size: ", X_test.shape)
```

```python
model = DecisionTreeClassifier(max_depth=5, random_state=42)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("Predictions:", y_pred[:10])
print("Actual:     ", list(y_test[:10]))
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc:.2%}")

# Confusion matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
# [[True Neg, False Pos],
#  [False Neg, True Pos]]
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import pandas as pd

# Confusion matrix plot
ConfusionMatrixDisplay(cm, display_labels=["No Disease","Disease"]).plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()

# Feature importance
importances = pd.Series(model.feature_importances_, index=X.columns)
importances.sort_values().plot(kind="barh", color="#3498DB")
plt.title("Feature Importance")
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

for name, clf in [
    ("Decision Tree",    DecisionTreeClassifier(max_depth=5)),
    ("Random Forest",    RandomForestClassifier(n_estimators=100)),
    ("Logistic Regression", LogisticRegression(max_iter=1000)),
]:
    clf.fit(X_train, y_train)
    acc = accuracy_score(y_test, clf.predict(X_test))
    print(f"{name:25s} → {acc:.2%}")
```
